## Data Cleaning and Preprocessing

```python
df.isnull().sum()
```

```python
# Filling the missing values
df['director']=df['director'].fillna('Unknown')
df['cast']=df['cast'].fillna('Unknown')
df['country']=df['country'].fillna('Unknown')
```

```python
df.isnull().sum()
```

```python
# Convert date
df['date_added'] = pd.to_datetime(df['date_added'])

# Extract year
df['year_added'] = df['date_added'].dt.year
```

```python
df.head()
```

```python
df.info()
```

```python
df.isnull().sum()
```

```python
df['date_added']=df['date_added'].fillna('Unknown')
df['rating']=df['rating'].fillna('Unknown')
df['year_added'] = df['year_added'].fillna(df['year_added'].mean())
```

```python
df.isnull().sum()
```

```python
numeric_stats=df.describe(include='number')
print(numeric_stats)
```

```python
categorial_cols=df.select_dtypes(include=['object','category']).columns
for col in categorial_cols:
    print(f"\n-- Value counts for {col} ---")
    print(df[col].value_counts())
```

```python
# let's visualize this details to understand in better way
import matplotlib.pyplot as plt
```

```python
# Comparing number of movies and Tv shows on Netflix
df['type'].value_counts().plot(kind='bar')
plt.title("Movies vs TV Shows")
plt.show()
```

```python
#content added over years 

df['year_added'].value_counts().sort_index().plot()
plt.title("Content Added Over Years")
plt.xlabel("Year")
plt.ylabel("Count")
plt.show()
```

```python
df['country'].value_counts().head(10).plot(kind='bar')
plt.title("Top 10 Content Producing Countries")
plt.show()
```

```python
df['rating'].value_counts().plot(kind='bar')
plt.title("Ratings Distribution")
plt.show()
```

```python
# Convert type to numeric
df['type_num'] = df['type'].map({'Movie': 0, 'TV Show': 1})

# Correlation
corr = df[['release_year', 'type_num']].corr()

print(corr)
```

```python
import seaborn as sns
sns.heatmap(corr, annot=True)
plt.show()
```

```python
df['duration'].value_counts().head(10).plot(kind='bar')
plt.title("Duration Distribution")
plt.show()
```

```python
# Top genres
df['genres'].str.split(',').explode().value_counts().head(10).plot(kind='bar')
plt.title("Top Genres")
plt.show()
```
